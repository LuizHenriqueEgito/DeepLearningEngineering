```python
import torch
import torch.nn.functional as F
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset
```

```python
from tqdm import tqdm
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap
```

```python
from utils import plot_decision_boundary, accuracy_fn
```

```python
# create data
X = torch.tensor([[0, 0], [0, 1], [1, 0], [1, 1]], dtype=torch.float32)
y = torch.tensor([[0], [1], [1], [0]], dtype=torch.float32)
```

```python
data = torch.cat((X, y), 1)
```

```python
cmap_ = ListedColormap(['#ee3737', '#198bd9'])
```

```python
# plotting the data
fig, ax = plt.subplots()
ax.scatter(x=data[:, 0], y=data[:, 1], c=data[:, 2], s=250,cmap=cmap_, edgecolors='k')
ax.set_title('XOR plot')
ax.set_xlim(-0.2, 1.2)
ax.set_ylim(-0.2, 1.2)
ax.grid(True, linestyle='dotted')
plt.show()
```

```python
xor_model = nn.Sequential(
    nn.Linear(in_features=2, out_features=4),
    nn.ReLU(),
    nn.Linear(4, 1),
    nn.Sigmoid()
)
```

```python
xor_model
```

```python
# make a simple model
class XORModel(nn.Module):
    def __init__(self):
        super(XORModel, self).__init__()
        self.layer_in = nn.Linear(in_features=2, out_features=4)
        self.layer_hidden = nn.Linear(4, 1)
        self.relu = nn.ReLU()
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        x = self.relu(self.layer_in(x))
        x = self.sigmoid(self.layer_hidden(x))
        return x

# Instanciando o modelo
xor_model = XORModel()
```

```python
# prep the dataset using TensorDataset
dataset = TensorDataset(X, y)
# Prep DataLoader
dataloader = DataLoader(dataset, batch_size=4, shuffle=False)
```

```python
# optmize model, i will put the lr very low for see the model's training
optimizer = optim.SGD(xor_model.parameters(), lr=0.5)
```

```python
EPOCHS = 1000
for epoch in tqdm(range(EPOCHS)):
    xor_model.train()
    running_loss = 0
    running_accuracy = 0
    for x_train, y_train in dataloader:
        optimizer.zero_grad()
        outputs = xor_model(x_train)
        
        loss = F.binary_cross_entropy(outputs, y_train)
        loss.backward()
        optimizer.step()
        running_loss += loss.item()
        
        predicted_labels = (outputs >= 0.5).float()
        accuracy = (predicted_labels == y_train).float().mean().item()
        running_accuracy += accuracy
    
    if epoch %5 == 0:
        plot_decision_boundary(model=xor_model, s=500, color_map=cmap_ ,dataset=dataset, cols=[0, 1])
        print(f'Época: {epoch+1} | Loss: {running_loss / len(dataloader)} | acc: {running_accuracy / len(dataloader)}')
```

```python
import matplotlib.pyplot as plt
from matplotlib.widgets import Button
import os

class ImageBrowser:
    def __init__(self, image_folder):
        self.image_folder = image_folder
        self.image_files = sorted([f for f in os.listdir(image_folder) if f.endswith('.png')])
        self.current_index = 0

        self.fig, self.ax = plt.subplots()
        plt.subplots_adjust(bottom=0.2)
        self.img = plt.imread(os.path.join(image_folder, self.image_files[self.current_index]))
        self.ax.imshow(self.img)

        self.next_button_ax = plt.axes([0.81, 0.05, 0.1, 0.075])
        self.next_button = Button(self.next_button_ax, 'Next')
        self.next_button.on_clicked(self.next_image)

        self.prev_button_ax = plt.axes([0.7, 0.05, 0.1, 0.075])
        self.prev_button = Button(self.prev_button_ax, 'Previous')
        self.prev_button.on_clicked(self.prev_image)

    def next_image(self, event):
        self.current_index = (self.current_index + 1) % len(self.image_files)
        self.update_image()

    def prev_image(self, event):
        self.current_index = (self.current_index - 1) % len(self.image_files)
        self.update_image()

    def update_image(self):
        self.img = plt.imread(os.path.join(self.image_folder, self.image_files[self.current_index]))
        self.ax.imshow(self.img)
        self.fig.canvas.draw()

# Exemplo de uso:
image_folder = 'images'
browser = ImageBrowser(image_folder)
plt.show()
```
