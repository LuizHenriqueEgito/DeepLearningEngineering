```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
```

```python
SEED = 42
```

```python
URI_DATASET = (
    "https://archive.ics.uci.edu/ml/"
    "machine-learning-databases/00350/"
    "default%20of%20credit%20card%20clients.xls"
)

df = pd.read_excel(URI_DATASET, header=1)

df.head()
```

```python
df = df.rename(
    columns={
        "default payment next month": "label"
    }
)
print(df.shape)
```

```python
df_train, df_test = train_test_split(df, test_size=0.2, shuffle=True, random_state=SEED)
```

```python
df_train.to_parquet('dataset/df_train.pq')
df_test.to_parquet('dataset/df_test.pq')
```

```python



# meses do dataset
months = [
    ("1", "PAY_0"),
    ("2", "PAY_2"),
    ("3", "PAY_3"),
    ("4", "PAY_4"),
    ("5", "PAY_5"),
    ("6", "PAY_6"),
]


sequences = []


for month, pay_col in months:

    month_features = df[
        [
            f"BILL_AMT{month}",  # valor da fatura
            f"PAY_AMT{month}",   # valor pago
            pay_col,             # status de pagamento: são os meses de atraso 0 está em dia, 1, 2, 3 meses em atraso
            "LIMIT_BAL"          # limite de crédito
        ]
    ].values

    sequences.append(month_features)



# transforma em tensor 3D:
# (clientes, meses, features)

X = np.stack(
    sequences,
    axis=1
)


print(X.shape)
```
